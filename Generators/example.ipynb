### Generators

Generators are simple way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
#example

def gen_func(n:int):
    for i in range(n):
        return i**2
```

```python
gen_func(3)
```

```python
#using yield keyword

def gen_func(n:int):
    for i in range(n):
        yield i**2
```

```python
gen_func(3)
```

```python
func_iterator = gen_func(3)
```

```python
#iterate over one by one using next

next(func_iterator)
```

```python
next(func_iterator)
```

```python
next(func_iterator)
```

```python
next(func_iterator)
```

```python
# to get all elements at once

for i in func_iterator:
    print(i)
```
