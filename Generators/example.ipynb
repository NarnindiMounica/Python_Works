### Generators

Generators are simple way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
#example

def gen_func(n:int):
    for i in range(n):
        return i**2
```

```python
gen_func(3)
```

```python
#using yield keyword

def gen_func(n:int):
    for i in range(n):
        yield i**2
```

```python
gen_func(3)
```

```python
func_iterator = gen_func(3)
```

```python
#iterate over one by one using next

next(func_iterator)
```

```python
next(func_iterator)
```

```python
next(func_iterator)
```

```python
next(func_iterator)
```

```python
# to get all elements at once

for i in func_iterator:
    print(i)
```

```python
#multiple values can be yield

def multi_yield_func():
    yield 1
    yield 2
    yield 3
    
```

```python
multi_yield_gen = multi_yield_func()
multi_yield_gen
```

```python
next(multi_yield_gen)
```

```python
next(multi_yield_gen)
```

```python
next(multi_yield_gen)
```

```python
next(multi_yield_gen)
```

```python
multi_yield_gen = multi_yield_func()
```

```python
#to get all values at once

for value in multi_yield_gen:
    print(value)
```

##### Practical Example: Reading Large Files

Generators are particularly useful for reading lage files because they allow you to process one line at a time without loading  the entire file into memory.

```python
#example: reading large file

def file_read(file_path:str):
    with open(file_path, "r") as file_obj:
        for line in file_obj.readlines():
            yield line
```

```python
for i in file_read("data.txt"):
    print(i)
    print(200*"---")
```
