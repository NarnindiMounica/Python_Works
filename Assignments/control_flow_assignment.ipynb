```python
### Assignment 1: Simple if Statement

#Write a program that asks the user to input a number and prints whether the number is positive.

user_num = int(input("Enter a number: "))
if user_num >= 0 :
    print(f"Number {user_num} is positive")
else:
    print(f"Number {user_num} is negative")    
```

```python
### Assignment 2: if-else Statement

#Write a program that asks the user to input a number and prints whether the number is positive or negative.

user_num = int(input("Enter a number: "))
if user_num >= 0 :
    print(f"Number {user_num} is positive")
else:
    print(f"Number {user_num} is negative")  
```

```python
### Assignment 3: if-elif-else Statement

#Write a program that asks the user to input a number and prints whether the number is positive, negative, or zero.

user_num = int(input("Enter a number: "))
if user_num > 0 :
    print(f"Number {user_num} is positive")
elif user_num < 0:
    print(f"Number {user_num} is negative")
else:
    print(f"Number {user_num} is zero") 
```

```python
### Assignment 4: Nested if Statement

#Write a program that asks the user to input a number and prints whether the number is positive and even, positive and odd, or negative.

user_num = int(input("Enter a number: "))
if user_num > 0 :
    if user_num % 2 == 0:
        print(f"Number {user_num} is positive and even")
    else:
        print(f"Number {user_num} is positive and odd") 
else:
    print(f"Number {user_num} is negative")           
```

```python
### Assignment 5: for Loop

#Write a program that prints all the numbers from 1 to 10 using a for loop.

for i in range(1,10):
    print(i)
```

```python
### Assignment 6: while Loop

#Write a program that prints all the numbers from 1 to 10 using a while loop.

i = 1
while i <= 10:
    print(i)
    i = i + 1
```

```python
### Assignment 7: Nested Loops

#Write a program that prints a 5x5 grid of asterisks (*) using nested loops.

for i in range(5):
    for j in range(5):
        print('*', end=" ")
    print("\n")    
```

```python
### Assignment 8: break Statement

#Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.

user_input = 1
sum = 0
while user_input != 0:
    sum = sum +user_input
    user_input = int(input("enter a number:"))
    if user_input == 0:
        print("ok bye")
        break
print(sum-1)    
```

```python
### Assignment 9: continue Statement

#Write a program that prints all the numbers from 1 to 10 except 5 using a for loop and continue statement.

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
### Assignment 10: pass Statement

#Write a program that defines an empty function using the pass statement.

def empty_func():
    pass

empty_func()
```

```python
### Assignment 11: Combining Loops and Conditionals

#Write a program that asks the user to input a number and prints all the even numbers from 1 to that number using a for loop.

user_int = int(input("enter a number:"))

for i in range(user_int+1):
    if i % 2 == 0:
        print(i)
```

```python
### Assignment 12: Factorial Calculation

#Write a program that calculates the factorial of a number input by the user using a while loop.

user_int = int(input("enter a number:"))
fact_num = 1
while user_int != 1:
    fact_num = fact_num * user_int 
    user_int = user_int - 1
print(fact_num)    
```
