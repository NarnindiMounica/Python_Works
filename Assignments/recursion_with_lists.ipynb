### Recursion with Lists/Arrays

```python
#write a function to check if given list is sorted or not

def check_sort(lst):

    if len(lst) in [0, 1]:
        return True
    elif lst[0]>lst[1]:
        return False
    else:
       return check_sort(lst[1:])        

check_sort(lst=[20,50,70,100,120])
```
