```python
#write a function to check if given list is sorted or not

def check_sort(lst):

    if len(lst) in [0, 1]:
        return True
    elif lst[0]>lst[1]:
        return False
    else:
       return check_sort(lst[1:])        

check_sort(lst=[20,50,70,100,120])
```

```python
#write code to fins sum of arrays using recursion

def recursive_sum(arr):
    """
    Function to calculate the sum of an array using recursion.
    
    Parameters:
    arr (list of int): The array of integers.
    
    Returns:
    int: The sum of the array elements.
    """
    # Your code here
    if len(arr) == 0:
        return 0
    else:
        sum_value = recursive_sum(arr[1:]) + arr[0]
        return sum_value

recursive_sum(arr=[1,2,3,4,5])
```

```python
#write a code to find first and last index of given element from list using recursion.

def first_index_of_element(arr, element):
    if len(arr) == 0 :
        return -1
    elif arr[0] == element:
        return 0
    else: 
        ans_from_list = first_index_of_element(arr[1:], element)
        if ans_from_list == -1:
            return -1 
        else: 
            return ans_from_list +1



first_index_of_element(arr=[3,2,1,3,2,4,5,6], element=10)
```

```python
#write a code to find first and last index of given element from list using recursion.

def first_index_of_element(arr, element):
    if len(arr) == 0 :
        return -1
    elif arr[-1] == element:
        return len(arr)-1
    else: 
        return first_index_of_element(arr[:-1], element)



first_index_of_element(arr=[3,2,1,3,2,4,5,6], element=10)
```

```python
#write a code to print all indexes of an element using recursion

def print_elements(arr, element, inx):
    if len(arr)==inx:
        return -1
    elif arr[inx]==element:
        print(inx) 
    print_elements(arr, element, inx+1)
                    

print_elements(arr=[3,2,1,2,3,4,2,5], element=2, inx=0)
```
