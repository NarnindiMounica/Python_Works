### Recursion with strings

```python
#removing given character from a string

def remove_character(string, char):
    if len(string)==0 or string == "":
        return string

    small_answer = remove_character(string[1:], char)

    if string[0]!=char:
        return string[0]+small_answer
    else:
        return  small_answer


remove_character(string="monicaaaaaaaa", char='a')
```

```python
#write code to check if string is palindrome or not
def check_palindrome_helper(string, start, end):
    if start >= end:
        return True

    if string[start] != string[end] :
        return False
    else:
        return check_palindrome_helper(string, start+1, end-1)     

def check_palindrome(string):

    return check_palindrome_helper(string, start=0, end=len(string)-1)

check_palindrome(string="abcbaaa")
```
