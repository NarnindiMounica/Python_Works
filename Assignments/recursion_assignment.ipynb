### Recursion Problems Assignment

```python
#Factorial of a number with recursion

def factorial_of_num(num):

    if num==1 or 0:
        return 1
    else:
        return num * factorial_of_num(num -1)
    
factorial_of_num(num = 5)    
    
```

```python
#Sum of n numbers

def sum_of_nums(num):

    if num == 0:
        return 0
    elif num == 1:
        return 1
    else:
        return num + sum_of_nums(num -1)
    
sum_of_nums(num = 5)    
```

```python
#Number of digits

def num_of_digits(num):
    count = 0
    if num >= 1 and num <= 9:
        return 1
    else:
        num = int(num/10)
        count = count + 1

        return count + num_of_digits(num)
    
num_of_digits(num=12367800)    
```

```python
#Fibonacci Series Using Recursion --fibonnaci(5)=0,1,1,2,3,5

def fibonacci(num):

     if num == 0:
          return 0
     elif num == 1:
          return 1
     else:
          return fibonacci(num-1)+fibonacci(num-2)
     
fibonacci(num=5)     
```

```python
#Print 1 to N using Recursion
'''Problem Description:

You are given a positive integer n. Your task is to return a list of integers from 1 to n using recursion.'''
lst = []
def print1ton(num):
    if num == 0:
        return lst
    else:

        lst.append(num)
        print1ton(num-1)
    return lst   
    
print1ton(num=5)
```
