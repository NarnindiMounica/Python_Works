### Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code.

```python
#function copy

def welcome():
    return "welcome to this notebook"
```

```python
wel = welcome()
wel
```

```python
wel_2 = wel
wel_2
```

```python
del wel
wel_2
```

```python
#closures: function inside a function

def main_welcome():
    msg = "welcome to main"
    def sub_welcome_method():
        print("this is sub_welcome method first line ")
        print("this is sub_welcome method last line")
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
#using main method variable in sub method
def main_welcome():
    msg = "welcome to main"
    def sub_welcome_method():
        print("this is sub_welcome method first line ")
        print(msg)
        print("this is sub_welcome method last line")
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
#using function passed as parameter in sub method
def main_welcome(func, msg):
    
    def sub_welcome_method():
        print("this is sub_welcome method first line ")
        func(msg)
        print("this is sub_welcome method last line")
    return sub_welcome_method()

```

```python
main_welcome(print, "this is passed in runtime")
```

```python
#decorator

def main_welcome(func):
    
    def sub_welcome_method():
        print("this is sub_welcome method first line ")
        func()
        print("this is sub_welcome method last line")
    return sub_welcome_method()
```

```python
def welcome_course():
    print("course welcome")
```

```python
main_welcome(welcome_course)
```

```python
@main_welcome
def welcome_course():
    print("course welcome")
```

```python
#decorators with arugments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper        
    return decorator    
```

```python
@repeat(3)
def say_hello():
    print("hello!!")
```

```python
say_hello()
```

```python
def main_welcome():
    def sub_welcome(func):
        def wrapper(*args, **kwargs):
            func(*args, **kwargs)
        return wrapper
    return sub_welcome        
```

```python
@main_welcome()
def welcome_course(msg):
    print(msg)

welcome_course("welcome decorator")
```
