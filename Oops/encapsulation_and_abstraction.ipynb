### Access Modifiers or Access Variables

* Public
* Private
* Protected

```python
#examples

class Person:

    def __init__(self, name, age, gender):
        self.name = name #public - can be accessed from anywhere
        self.__age = age #private - can be accessed only inside class
        self._gender = gender #protected - can be accessed in class and it's derived classes

class Student(Person):

    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

    def wish_student(self):
        return f"Hello {self.name}"

    def age_restrict(self):
        return  f"{self._Person__age}"  #data mangling

    def gender_reveal(self):
        return f"{self._gender}"               

student_obj = Student(name="moni", age=25, gender="female")
print(student_obj.wish_student())
print(student_obj.age_restrict())
print(student_obj.gender_reveal())
```

### Encapsulation using getter & setter 

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def set_name(self, name):
        self.__name = name

    def set_age(self, age):
        self.__age = age


student_obj = Student(name="moni", age=25)
student_obj.set_name(name="rita")
student_obj.set_age(age=24)
print(student_obj.get_name())
print(student_obj.get_age())
```

### Abstraction

Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and efforts.

```python
#Example

from abc import ABC, abstractmethod

class Vehicle(ABC):

    def start_vehicle(self):
        return "The vehicle has started"

    def start_engine(self):
        pass

class Car(Vehicle):

    def start_engine(self):
        return "Car engine started"


car_obj = Car()
print(car_obj.start_engine())
print(car_obj.start_vehicle())        

```
