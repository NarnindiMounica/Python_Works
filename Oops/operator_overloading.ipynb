### Operator Overloading

```python
#example

class Operations:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, vector):
        return self.x + vector.x, self.y + vector.y

    def __sub__(self, vector):
        return self.x - vector.x, self.y - vector.y    


vect1 = Operations(x=2, y=3)
vect2 = Operations(x=1, y=4)

print(vect1 + vect2)
print(vect1-vect2)
```

```python
#example

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)    


v1 + v2
```

```python
#Complex Number Class
'''Problem Description

Design a Python class named ComplexNumber to represent complex numbers and implement common methods to work with them. A complex number is a number of the form a+bia + bia+bi, where aaa is the real part and bbb is the imaginary part, and iii is the imaginary unit (i.e., −1\sqrt{-1}−1​).'''

class ComplexNumber:
    def __init__(self, real, imaginary):
        # Initialize real and imaginary parts
        self.real = real
        self.imaginary = imaginary
    
    def add(self, other):
        # Add another ComplexNumber object
        
        rea = int(self.real) + int(other.real)
        ima = int(self.imaginary) + int(other.imaginary)
        return ComplexNumber(rea, ima)
    
    def subtract(self, other):
        # Subtract another ComplexNumber object
        rea = int(self.real) - int(other.real)
        ima = int(self.imaginary) - int(other.imaginary)
        return ComplexNumber(rea, ima)
    
    def multiply(self, other):
        # Multiply by another ComplexNumber object
        rea = (int(self.real) * int(other.real)) - (int(self.imaginary)* int(other.imaginary))
        ima = (int(self.real)* int(other.imaginary)) + (int(self.imaginary)* int(other.real))
        
        return ComplexNumber(rea, ima)
    
    def __eq__(self, other):
        # Compare two ComplexNumber objects
        rea = self.real == other.real
        ima = self.imaginary == other.imaginary
        return rea and ima
    
    def __str__(self):
        # Return string representation of the complex number
        self.real = int(self.real)
        self.imaginary = int(self.imaginary)
        if self.imaginary > 0:
            return f"{self.real} + {self.imaginary}i"
        else:
            return f"{self.real} - {abs(self.imaginary)}i"


complex_num1 = ComplexNumber(1,2)
complex_num2 = ComplexNumber(3,4)
print(complex_num1.add(complex_num2))
```
