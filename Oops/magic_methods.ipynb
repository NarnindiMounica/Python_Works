### Magic Methods / Dunder Methods:

Magic methods are predefined methods in Python that you can override to change the behavior of your objects.


\_\_init__: initializes a new instance of a class.  
\_\_str__: returns a string representation of an object.  
\_\_repr__: returns an official string representation of an object.  
\_\_len__: returns the length of an object.  
\_\_getitem__: gets an item from a container.  
\_\_setitem__: sets an item in a container.  


```python
#example

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is new in class"

    def __repr__(self):
        return f"Student(name={self.name}, age={self.age})"


student_obj = Student(name="moni", age=27)
print(student_obj) 
print(repr(student_obj))       
```
