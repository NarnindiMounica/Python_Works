### Inheritance

Inheritance is fundamental concept in Object oriented programming (oops) that allows a class to inherit attributes and methods from other class.

```python
#example

class Car:

    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")    
        
```

```python
cars = Car(windows=4, doors=5, enginetype="petrol")
cars.drive()
```

```python
#inherited class

class Tesla(Car):

    def __init__(self, windows, doors, enginetype, self_driving):
        super().__init__(windows, doors, enginetype)
        self.self_driving = self_driving

    def is_self_driving(self):
        print(f"Tesla supports self driving: {self.self_driving}")
```

```python
tesla1 = Tesla(windows=4, doors=5, enginetype="petrol", self_driving="yes")
tesla1.drive()
tesla1.is_self_driving()
```

```python
#Multiple Inheritance - inheriting from more than one class

class Animal:
    def __init__(self, name):
        self.name = name

class Pet:
    def __init__(self, owner):
        self.owner = owner

class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal().__init__(self, name)
        Pet().__init__(self, owner)

    def calling(self):
        return f"{self.name} is a pet owned by {self.owner}"                


```
