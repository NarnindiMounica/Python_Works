### Inheritance

Inheritance is fundamental concept in Object oriented programming (oops) that allows a class to inherit attributes and methods from other class.

```python
#example

class Car:

    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")    
        
```

```python
cars = Car(windows=4, doors=5, enginetype="petrol")
cars.drive()
```
