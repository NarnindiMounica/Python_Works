### Custom Exception

```python
class dobException(Exception):

    def error_message(self):
        print("this is dob exception")

try:
    dob_year = int(input("Enter year of birth: "))
    age = 2026 - dob_year

    if age in range(20, 30):
        print("valid age")
    else:
        raise dobException
except dobException as e:
    e.error_message()
```
