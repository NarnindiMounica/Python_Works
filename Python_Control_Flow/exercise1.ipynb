```python
#write a program that asks user to input numbers until they input 0, the program should print the sum of all input numbers.

sum=0
while True:
    number = int(input('Enter a number: '))
    if number != 0:
        sum = sum+number
    else:
        break
print('sum of input numbers', sum)            
        
```

```python
#write a program that calculates the factorial of a number input by user using a while loop

fact_num = int(input('enter a number:'))
if fact_num==0 or fact_num==1:
    print(1)
else:
    i=1
    fact=1
    while i < fact_num:
        fact= fact * (i+1)
        i = i+1
    print(fact)    

                     
```

```python
#write a program that checks if a number input by a user is a prime number using a for loop
number = int(input('enter a number: '))
if number in (1,0):
    count=0
else:
    count=0    
    for i in range(1,number+1):
        if number%i == 0:
            count +=1
        print(count)    
if count == 2:
    print('number is prime')
else:
    print('number is not prime')    

        
```
