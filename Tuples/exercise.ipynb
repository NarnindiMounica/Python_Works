```python
#create a tuple with duplicate elements and count the occurences of an element. Find the index of first occurence of an element in the tuple.count
sample = (1,2,3,3,2,1,4,2,3,4)
for element in set(sample):
    print(element, 'count :', sample.count(element))
    print(element, 'first occurence at:', sample.index(element))
```
