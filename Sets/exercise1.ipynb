```python
#create a set with first 10 positive integers, add 11 and remove 1 from it.
my_set = set(range(1,11))
my_set.add(11)
my_set.remove(1)
print(my_set)
```

```python
#create two sets: one with the first 5 positive integers and another with first 5 even integers. Perform and print the results of union, intersection, difference and symmetric difference operations on sets.

#symmetric difference= set of elements in either sets but not in their intersection.
set1 = set(range(1,6))
set2 = set(range(2,11,2))
print(set1 | set2)
print(set1 & set2)
print(set1 - set2)
print(set1 ^ set2)
```
