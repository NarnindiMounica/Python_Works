```python
#Write a function that takes two integers as input and returns their division. Use try, except and finally blocks to handle division by zero and print an appropriate message.

def division(num1, num2):
    try:
        return num1/num2
    except Exception as e:
       return e
    finally:
        print('operation performed')    


print(division(1,4))
print(division(2,0))
```

```python
#write a function that takes a list of integers and returns their sum. use try, except and finally blocks to handle TypeError if a non-integer value is encountered and print an appropriate message.

def list_sum(int_list: list):
    sum=0
    try:
        for item in int_list:
           sum += item
    except TypeError:
        print('non-int value passed in list')
        sum=None
    finally:
        print('operation completed') 
    return sum       


list_sum([1,2,3,4,5,6,'a']) 

list_sum([1,2,3,4,5,6])          
    
  
```

```python
#Write a function that prompts the user to enter an integer. Use try, except and finally blocks to handle ValueError if the user enters a non-integer value and print an appropriate message.

def display_int():
    try:
        number= int(input('enter an integer: ')   )
    except ValueError:
        print('only integers allowed')
        number=None
    finally:
        print('Done')
    return number    

print(display_int())
display_int()
```

```python
#write a function that takes a dictionary and a key as input and returns the value associated with the key. Use try, except and finally blocks to handle KeyError if the key is not found in the dictionary and print an appropriate message.
def key_val(dictionary, key):
    try:
        value= dictionary[key]
    except KeyError:
        print('key not in given dictionary')
        value=None
    finally:
        print('done')
    return value        

print(key_val({1:2, 3:4, 5:6}, 5))
print(key_val({1:2, 3:4, 5:6}, 9))
```

```python
#Write a function that performs nested exception handling. It should first attempt to convert a string to an integer, and then attempt to divide by that integer. Use nested try, except and finally blocks to handle ValueError and ZeroDivisionError and print appropriate messgaes.
def nested_exception(s:str):
    try:
        try:
            number=int(s)
        except ValueError as e:
            print('not an integer : ', e)
            number=None
        finally:
            print('type casting completed')  

        division=10/number    
    except ZeroDivisionError as e:
        print("can't divide by zero: ", e)
        division=None
    except Exception as e:
        print(e)
        division=None    
    return division   

print(nested_exception('4') )
print(nested_exception('0'))
print(nested_exception('a'))



```

```python
#write a function that takes a list and an index as input and returns the element at the given index. use try, except and finally blocks to handle indexError if the index is out of range and print an appropriate message.

def index_val(my_list:list, ind:int):
    try:
        value=my_list[ind]
    except IndexError as e:
        print(e) 
        value=None 
    finally:
        print('task completed')  

    return value      

print(index_val([1,2,3],2))
print(index_val([1,2,3],3))
```

```python
#write a function that attempts to parse a JSON string. Use try, except and finally blocks to handle JSONDecodeError if the string is not a valid Json and print an appropriate message.

import json
def json_conv(string):
    try:
        json_string= json.loads(string)
        return json_string
    except json.JSONDecodeError as e:
        print(e)
        return None
    
print(json_conv('"{\'name\':\'moni\',\'age\':28}"'))
print(json_conv('json'))
```

```python
#Define a custom exception named 'NegativeNumberError'. write a function that raises this exception if a negative number is encountered in a list. Use try, except and finally blocks to handle the custom exception and print an appropriate messgae.

class NegativeNumberError(Exception):
    pass

def check_for_negative(lst):
    try:
        for item in lst:
            if item < 0:
                raise NegativeNumberError(f'Negative number found {item}')
    except  NegativeNumberError as e:
        print(e)
    finally:
        print('execution completed.')           

check_for_negative([1,2,3,0,-1])
```

```python
#write a function that calls another function which may raise an exception. Use try, except and finally blocks to handle the exception and print an appropriate message.
class RiskError(Exception):
    pass

def risky_fun():
    raise RiskError(f'an error occurred in this function')

def safe_fun():
    try:
        risky_fun()
    except RiskError as e:
        print(e)
    finally:
        print('risky function is called') 


safe_fun()

```
