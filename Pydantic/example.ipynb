### Pydantic

```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):

    name:str
    age:int
    city:str
```

```python
persona = Person(name="Moni", age=27, city="pune")
persona
```

```python
persona.name
```

```python
persona.name=32
```
