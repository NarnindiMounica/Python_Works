### Pydantic

```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):

    name:str
    age:int
    city:str
```

```python
persona = Person(name="Moni", age=27, city="pune")
persona
```

```python
persona.name
```

```python
persona.name=32
```

```python
persona
```

```python
#to validate datatype at the time of assignment we need to use as below

from pydantic import BaseModel, ConfigDict

class Person(BaseModel):
    name:str
    age:int
    city:str

    model_config = ConfigDict(validate_assignment=True)
```

```python
personb = Person(name="Moni", age=27, city="hyd")
personb
```

```python
personb.name=32
```

##### Model with Optional Fields

```python
from typing import Optional

class Employee(BaseModel):
    id:int
    name:str
    department:str
    salary:Optional[float]=None
    is_active:Optional[bool]=False
```

```python
emp_1 = Employee(id=123, name='Moni', department="AI")
```

```python
emp_1
```

- Optional[datatype]: indicates the filed can be None

- Default value (None or False): makes the filed optional

Required fileds must still be provided

Pydantic validates types even for optionla fileds when values are provided

##### Complex structure with nested model

```python
from pydantic import BaseModel

class Address(BaseModel):
    street:str
    city:str
    zip_code:str

class Customer(BaseModel):
    name:str
    id:int
    address:Address

customer_a = Customer(name="Moni", id=123, address=
{"street":"street no-124",
"city":"pune",
"zip_code":"8907"})      
```

#### Pydantic Fields: Customization and Constraints 

This Field function in Pydantic enhances model fields beyond basic type ints by allowing you to specify validation rules, default values, aliases and more

```python
from pydantic import BaseModel, Field

class Item(BaseModel):
    name:str=Field(min_length=2, max_length=50)
    price:float=Field(gt=0, le=1000)
    quantity:int=Field(ge=0)


item = Item(name="books", price=100, quantity=2)    
```

```python
item
```

```python
Item.schema()
```
