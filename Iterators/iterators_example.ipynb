### Iterators 

Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection seqentially without exposing the underlying struture.

```python
#example

mylist = [1, 2, 3, 4]
for i in mylist:
    print(i)
```

```python
#converting it into iterator object

iterator_obj = iter(mylist)
iterator_obj
```

```python
#looping one element through iterator

next(iterator_obj)
```

```python
iterator_obj = iter(mylist)
```

```python
#using try and except block to handle exception

try: 
    print(next(iterator_obj))
except StopIteration:
    print("No elements left to iterator over..")
    
```
