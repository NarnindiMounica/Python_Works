```python
#create a defaultdict where each key has a default value of an empty list. Add some elements to the list and print the dictionary.

from collections import defaultdict

my_dict = defaultdict(list)
my_dict['a'].append(1)
my_dict['b'].append(2)
print(my_dict)
```

```python
#write a function that takes a string and returns a dictionary with the count of each character in the string. print the dictionary

def char_count(my_string):
    my_dict={}
    for i in set(my_string):
        my_dict[i]=my_string.count(i)
    return my_dict

my_string = 'my name is moni, and im a good girl'
print(char_count(my_string))
```
